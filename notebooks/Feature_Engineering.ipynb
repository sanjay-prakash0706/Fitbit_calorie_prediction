```python
fe_df = df.copy()
```

```python
X = fe_df.drop('Calories_Burned_kcal', axis=1)
y = fe_df['Calories_Burned_kcal']
```

```python
fe_df['Fat_Category'] = pd.cut(
    fe_df['Fat_Percentage'],
    bins=[-float('inf'), 7.99, 14.99, 19.99, 24.99, 29.99, float('inf')],
    labels=[
        'Extremely Lean / Very Low',
        'Athletic / Very Fit',
        'Fit',
        'Average',
        'High',
        'Very High'
    ]
)
```

```python
fe_df['Fat_Category'].value_counts()
```

```python
fe_df[['Fat_Percentage', 'Fat_Category']].head(20)
```

```python
fe_df['Fat_Category'].isnull().sum()
```

```python
fe_df['BMI_Category'] = pd.cut(
    fe_df['BMI'],
    bins=[-float('inf'), 18.49, 24.99, 29.99, float('inf')],
    labels=[
        'Underweight',
        'Normal',
        'Overweight',
        'Obesity'
    ]
)
```

```python
fe_df[['BMI', 'BMI_Category']].head(20)
```

```python
fe_df['Weekly_Workout_Hours'] = (
    fe_df['Session_Duration_hours']
    * fe_df['Workout_Frequency_days_per_week']
)
```
