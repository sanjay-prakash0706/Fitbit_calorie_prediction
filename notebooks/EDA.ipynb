Bivariate Analysis

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df['Session_Duration_hours'],
    df['Calories_Burned_kcal']
)

plt.xlabel('Session Duration (hours)')
plt.ylabel('Calories Burned (kcal)')
plt.title('Session Duration vs Calories Burned')

plt.show()

"""The scatter plot shows a positive relationship between session duration
 and calories burned. In general,"""
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df['Effective_MET'],
    df['Calories_Burned_kcal']
)

plt.xlabel('Effective MET')
plt.ylabel('Calories Burned (kcal)')
plt.title('Effective MET vs Calories Burned')

plt.show()
"""Higher exercise intensity is generally associated with higher calorie expenditure.
 However, there are
 some outliers where participants with lower effective MET values 
 burned more calories than those with higher effective MET values."""
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df['Avg_BPM'],
    df['Calories_Burned_kcal']
)

plt.xlabel('Average BPM')
plt.ylabel('Calories Burned (kcal)')
plt.title('Average BPM vs Calories Burned')

plt.show()

"""he scatter plot indicates a positive relationship between average heart rate 
and calories burned. Higher average heart rates are generally
 associated with higher calorie expenditure."""
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df['Weight_kg'],
    df['Calories_Burned_kcal']
)

plt.xlabel('Weight (kg)')
plt.ylabel('Calories Burned (kcal)')
plt.title('Weight vs Calories Burned')

plt.show()

"""considerable variation is observed at higher weight levels,
 suggesting that other factors such as session duration and
   workout intensity also influence calorie expenditure"""
```

```python
plt.figure(figsize=(7,5))

df.boxplot(
    column='Calories_Burned_kcal',
    by='Gender'
)

plt.xlabel('Gender')
plt.ylabel('Calories Burned (kcal)')
plt.title('Calories Burned by Gender')
plt.suptitle('')

plt.show()

"""The calorie expenditure distributions for male and female participants are broadly similar,
 with comparable median values and spread"""
```

```python
plt.figure(figsize=(7,5))

df.boxplot(
    column='Calories_Burned_kcal',
    by='Experience_Level'
)

plt.xlabel('Experience Level')
plt.ylabel('Calories Burned (kcal)')
plt.title('Calories Burned by Experience Level')
plt.suptitle('')

plt.show()

"""The boxplot shows a clear increasing trend in calorie expenditure across experience levels.
 Participants with higher experience levels generally have higher median calories burned and greater variability."""
```

```python
plt.figure(figsize=(9,6))

plt.scatter(
    df['Session_Duration_hours'],
    df['Calories_Burned_kcal'],
    s=df['Effective_MET'] * 10,
    alpha=0.4
)

plt.xlabel('Session Duration (hours)')
plt.ylabel('Calories Burned (kcal)')
plt.title('Session Duration and Effective MET vs Calories Burned')

plt.show()
```

#correlation and heat map 

```python
correlation = df.corr(numeric_only=True)['Calories_Burned_kcal'].sort_values(
    ascending=False
)

correlation

"""Effective MET has the strongest observed linear association with calories burned 
among the variables analyzed"""
```

```python
corr_matrix = df.select_dtypes(include='number').corr()

plt.figure(figsize=(12,8))

plt.imshow(corr_matrix, cmap='coolwarm', interpolation='none')

plt.colorbar()

plt.xticks(
    range(len(corr_matrix.columns)),
    corr_matrix.columns,
    rotation=90
)

plt.yticks(
    range(len(corr_matrix.columns)),
    corr_matrix.columns
)

plt.title('Correlation Matrix')

plt.tight_layout()
plt.show()

"""Calories burned are most strongly associated with workout intensity,
 followed by session duration and heart-rate-related measures."""
```
